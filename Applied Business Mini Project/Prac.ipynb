```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("car_dealership_sales_5000.csv")

# Convert Sale_Date to datetime
df["Sale_Date"] = pd.to_datetime(df["Sale_Date"], errors="coerce")

# =========================================================
# 1. FIRST FIVE RECORDS
# =========================================================
print("=" * 60)
print("1. FIRST FIVE RECORDS")
print("=" * 60)
display(df.head())


# =========================================================
# 2. LAST FIVE RECORDS
# =========================================================
print("\n" + "=" * 60)
print("2. LAST FIVE RECORDS")
print("=" * 60)
display(df.tail())


# =========================================================
# 3. DATASET SHAPE
# =========================================================
print("\n" + "=" * 60)
print("3. DATASET SHAPE")
print("=" * 60)
print("Rows and Columns:", df.shape)


# =========================================================
# 4. COLUMN NAMES
# =========================================================
print("\n" + "=" * 60)
print("4. COLUMN NAMES")
print("=" * 60)
print(df.columns.tolist())


# =========================================================
# 5. DATA TYPES
# =========================================================
print("\n" + "=" * 60)
print("5. DATA TYPES")
print("=" * 60)
print(df.dtypes)


# =========================================================
# 6. MISSING VALUES
# =========================================================
print("\n" + "=" * 60)
print("6. MISSING VALUES")
print("=" * 60)
print(df.isnull().sum())
print("Total Missing Values:", df.isnull().sum().sum())


# =========================================================
# 7. DUPLICATE RECORDS
# =========================================================
print("\n" + "=" * 60)
print("7. DUPLICATE RECORDS")
print("=" * 60)
print("Total Duplicate Records:", df.duplicated().sum())


# =========================================================
# 8. UNIQUE VALUES
# =========================================================
print("\n" + "=" * 60)
print("8. UNIQUE VALUES")
print("=" * 60)

categorical_columns = [
    "Brand",
    "Car_Model",
    "Car_Category",
    "Fuel_Type",
    "Transmission",
    "Region",
    "State",
    "City",
    "Payment_Method"
]

for column in categorical_columns:
    print(f"\n{column}")
    print("Number of Unique Values:", df[column].nunique())
    print("Unique Values:", df[column].unique())


# =========================================================
# 9. BASIC STATISTICS
# =========================================================
print("\n" + "=" * 60)
print("9. BASIC STATISTICS")
print("=" * 60)
display(df.describe())


# =========================================================
# 10. BAR CHART 1 - SALES BY CAR CATEGORY
# =========================================================
category_sales = df.groupby("Car_Category")["Sales"].sum().sort_values(
    ascending=False
)

plt.figure(figsize=(8, 5))
sns.barplot(x=category_sales.index, y=category_sales.values)
plt.title("Total Sales by Car Category")
plt.xlabel("Car Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print("Observation: SUVs generate the highest total sales among the car categories.")


# =========================================================
# 11. BAR CHART 2 - PROFIT BY REGION
# =========================================================
region_profit = df.groupby("Region")["Profit"].sum().sort_values(
    ascending=False
)

plt.figure(figsize=(8, 5))
sns.barplot(x=region_profit.index, y=region_profit.values)
plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print("Observation: The chart shows differences in total profit across the four regions.")


# =========================================================
# 12. HISTOGRAM 1 - SALES
# =========================================================
plt.figure(figsize=(8, 5))
plt.hist(df["Sales"], bins=20, edgecolor="black")
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("Observation: Most sales transactions are concentrated within a particular sales range, with fewer transactions at very high sales values.")


# =========================================================
# 13. HISTOGRAM 2 - PROFIT
# =========================================================
plt.figure(figsize=(8, 5))
plt.hist(df["Profit"], bins=20, edgecolor="black")
plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("Observation: Most transactions generate moderate profits, while fewer transactions generate very high profits.")


# =========================================================
# 14. BOX PLOT - SELLING PRICE
# =========================================================
plt.figure(figsize=(8, 5))
sns.boxplot(y=df["Selling_Price"])
plt.title("Box Plot of Selling Price")
plt.ylabel("Selling Price")
plt.tight_layout()
plt.show()

print("Observation: The box plot shows the spread of selling prices and highlights potential extreme values.")


# =========================================================
# 15. PIE CHART - SALES BY CAR CATEGORY
# =========================================================
plt.figure(figsize=(7, 7))

plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Sales Distribution by Car Category")
plt.tight_layout()
plt.show()

print("Observation: SUVs contribute the largest share of total sales among the car categories.")


# =========================================================
# 16. SCATTER PLOT - SELLING PRICE VS PROFIT
# =========================================================
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Selling_Price",
    y="Profit"
)

plt.title("Selling Price vs Profit")
plt.xlabel("Selling Price")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()

print("Observation: The scatter plot shows a positive relationship between selling price and profit, as higher-priced cars generally tend to generate higher profits.")


# =========================================================
# EDA COMPLETED
# =========================================================
print("\n" + "=" * 60)
print("EDA COMPLETED SUCCESSFULLY")
print("=" * 60)
```

```python
import pandas as pd
# Load dataset
df = pd.read_csv("car_dealership_sales_5000.csv")
# Select important numerical variables
numerical_columns = [
    "Selling_Price",
    "Discount",
    "Sales",
    "Profit"
]
# Calculate descriptive statistics
results = []
for column in numerical_columns:
    results.append({
        "Variable": column,
        "Mean": df[column].mean(),
        "Median": df[column].median(),
        "Mode": df[column].mode()[0],
        "Minimum": df[column].min(),
        "Maximum": df[column].max(),
        "Range": df[column].max() - df[column].min(),
        "Variance": df[column].var(),
        "Standard Deviation": df[column].std()
    })
# Create result table
stats_table = pd.DataFrame(results)
# Round values
stats_table = stats_table.round(2)
# Display table
display(stats_table)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Load dataset
df = pd.read_csv("car_dealership_sales_5000.csv")
# Select Profit
profit = df["Profit"]
# -------------------------------
# Histogram
# -------------------------------
plt.figure(figsize=(8, 5))
plt.hist(profit, bins=20, edgecolor="black")
plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.show()
# -------------------------------
# Box Plot
# -------------------------------
plt.figure(figsize=(8, 5))
sns.boxplot(y=profit)
plt.title("Box Plot of Profit")
plt.ylabel("Profit")
plt.show()
# -------------------------------
# Statistical Values
# -------------------------------
print("Mean:", round(profit.mean(), 2))
print("Median:", round(profit.median(), 2))
print("Minimum:", round(profit.min(), 2))
print("Maximum:", round(profit.max(), 2))
print("Standard Deviation:", round(profit.std(), 2))
print("Skewness:", round(profit.skew(), 2))
```
